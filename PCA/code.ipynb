# principal component analysis

```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
data=load_digits()
df=pd.DataFrame(data.data,columns=data.feature_names)
```

```python
df['label']=data.target
```

doing the train test split and training the model before pca

```python
x=df.iloc[:,:64]
y=df.iloc[:,64:]
```

```python
y
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
knn=KNeighborsClassifier(n_neighbors=7)
knn.fit(x_train,y_train)
```

```python
y_pr=knn.predict(x_test)
acc=accuracy_score(y_true=y_test,y_pred=y_pr)
acc
```

now i will apply pca to the data and then again train the model

```python
x=pd.DataFrame(data.data)
y=pd.DataFrame(data.target)
```

```python
pca=PCA(n_components=5)
x_pca=pca.fit_transform(x)
```

```python
x_pca.shape
```

```python
x_train,x_test,y_train,y_test=train_test_split(x_pca,y,test_size=0.2,random_state=42)
```

```python
knn=KNeighborsClassifier(n_neighbors=5)
knn.fit(x_train,y_train)
```

```python
y_pre=knn.predict(x_test)
acc=accuracy_score(y_true=y_test,y_pred=y_pr)
acc
```

```python
for i in range(1,65):
  pca=PCA(n_components=i)
  x_pca=pca.fit_transform(x)
  x_train,x_test,y_train,y_test=train_test_split(x_pca,y,test_size=0.2,random_state=42)
  knn=KNeighborsClassifier(n_neighbors=5)
  knn.fit(x_train,y_train)
  y_pre=knn.predict(x_test)
  acc=accuracy_score(y_true=y_test,y_pred=y_pre)
  print(acc)
```
